# 02. RFM Feature Engineering & Skewness Resolution

This notebook computes the customer-level Recency, Frequency, and Monetary (RFM) metrics from cleaned transactional data and analyzes distribution skewness before and after log1p transformation.

```python
import pandas as pd
import numpy as np
from src.data.cleaner import clean_retail_data
from src.data.rfm_builder import build_rfm_features, transform_log1p
```

## 1. Load Cleaned Transactions & Compute RFM

```python
raw_df = pd.read_csv('data/online_retail_II.csv')
clean_df = clean_retail_data(raw_df)
rfm_df = build_rfm_features(clean_df)
print('Customer RFM matrix shape:', rfm_df.shape)
rfm_df.head(10)
```

## 2. Evaluate Raw Skewness & Summary Statistics

```python
print('Raw RFM Summary Statistics:')
print(rfm_df.describe())
print('\nRaw Feature Skewness:')
print(rfm_df.skew())
```

## 3. Apply Log1p Transformation & Measure Variance Stabilization

```python
log_rfm = transform_log1p(rfm_df)
print('Log-transformed RFM Summary Statistics:')
print(log_rfm.describe())
print('\nLog-transformed Feature Skewness:')
print(log_rfm.skew())
```
