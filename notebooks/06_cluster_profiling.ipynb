# 06. Cluster Profiling & Customer Persona Mapping

This notebook profiles the 5 optimal K-Means customer segments and maps them to commercial personas: Champions, Loyal Customers, Potential Loyalists, At Risk, and Hibernating.

```python
import pandas as pd
from sklearn.cluster import KMeans
from src.data.cleaner import clean_retail_data
from src.data.rfm_builder import build_rfm_features
from src.models.pipeline import build_rfm_preprocessor
from src.models.profiler import assign_personas, build_cluster_profiles
```

## 1. Fit Final K-Means Model (K=5)

```python
raw_df = pd.read_csv('data/online_retail_II.csv')
clean_df = clean_retail_data(raw_df)
rfm_df = build_rfm_features(clean_df)
preprocessor = build_rfm_preprocessor(scaler_type='standard')
X_scaled = preprocessor.fit_transform(rfm_df)

kmeans_model = KMeans(n_clusters=5, random_state=42, n_init=10)
cluster_labels = kmeans_model.fit_predict(X_scaled)
rfm_segmented = assign_personas(rfm_df, cluster_labels)
rfm_segmented.head(10)
```

## 2. Generate Commercial Cluster Profiles

```python
profiles = build_cluster_profiles(rfm_segmented)
print('Final Customer Segment Profiles:')
print(profiles.to_string(index=False))
```
