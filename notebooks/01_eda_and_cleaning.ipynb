# 01. Exploratory Data Analysis & Data Cleaning

This notebook analyzes the raw Online Retail II transactions (2009-2011), documents key data quirks (missing Customer IDs, invoice cancellations, non-positive quantities and prices), and validates the cleaning pipeline.

```python
import pandas as pd
import numpy as np
from src.data.cleaner import clean_retail_data
```

## 1. Load Raw Transactions

```python
raw_df = pd.read_csv('data/online_retail_II.csv')
print('Raw shape:', raw_df.shape)
raw_df.head()
```

## 2. Quantify Data Quirks & Missingness

```python
missing_stats = pd.DataFrame({
    'Null Count': raw_df.isnull().sum(),
    'Null Percentage (%)': (raw_df.isnull().sum() / len(raw_df)) * 100
})
print(missing_stats)
```

```python
cancellations = raw_df['Invoice'].astype(str).str.upper().str.startswith('C').sum()
neg_quantities = (raw_df['Quantity'] <= 0).sum()
non_pos_price = (raw_df['Price'] <= 0).sum()
print('Cancellations count:', cancellations)
print('Negative or zero quantity count:', neg_quantities)
print('Non-positive price count:', non_pos_price)
```

## 3. Execute Production Cleaning Pipeline

```python
clean_df = clean_retail_data(raw_df)
print('Cleaned dataset shape:', clean_df.shape)
print('Retained records ratio:', f'{len(clean_df) / len(raw_df):.2%}')
print('Distinct customers:', clean_df['Customer ID'].nunique())
clean_df.head()
```

## 4. Geographic & Transaction Distributions

```python
country_counts = clean_df['Country'].value_counts().head(10)
print('Top 10 Countries by Transaction Count:')
print(country_counts)
```
