# 05. Clustering Benchmark: K-Means vs Agglomerative vs DBSCAN

This notebook benchmarks K-Means against Agglomerative Hierarchical Clustering and DBSCAN density clustering across Silhouette Score, Davies-Bouldin Index, Calinski-Harabasz score, and inference latency.

```python
import pandas as pd
from src.data.cleaner import clean_retail_data
from src.data.rfm_builder import build_rfm_features
from src.models.pipeline import build_rfm_preprocessor
from src.models.benchmark import run_clustering_benchmark
```
