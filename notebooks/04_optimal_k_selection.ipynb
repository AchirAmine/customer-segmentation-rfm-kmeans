# 04. Optimal K Selection: Elbow Method, Silhouette & Davies-Bouldin

This notebook evaluates candidate cluster counts K in [2, 8] using Inertia (Elbow Method), Silhouette Score, and Davies-Bouldin Index on normalized RFM customer vectors.

```python
import pandas as pd
import numpy as np
from src.data.cleaner import clean_retail_data
from src.data.rfm_builder import build_rfm_features
from src.models.pipeline import build_rfm_preprocessor
from src.models.clustering_eval import evaluate_clustering_k
```

## 1. Prepare Standardized Customer RFM Features

```python
raw_df = pd.read_csv('data/online_retail_II.csv')
clean_df = clean_retail_data(raw_df)
rfm_df = build_rfm_features(clean_df)
preprocessor = build_rfm_preprocessor(scaler_type='standard')
X_scaled = preprocessor.fit_transform(rfm_df)
print('Processed matrix shape:', X_scaled.shape)
```

## 2. Compute Clustering Diagnostics Across K in [2, 8]

```python
k_metrics = evaluate_clustering_k(X_scaled, k_range=range(2, 9))
print('Clustering Evaluation Metrics by K:')
print(k_metrics.to_string(index=False))
```
